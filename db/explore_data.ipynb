# Explore Loan Data

Load a small sample from the raw CSV and inspect its structure.

```python
import pandas as pd

# Only reads the first 5 rows — instant, no need to load the whole file
df = pd.read_csv(r"C:\Users\Lenovo\credit-risk-platform\data\raw\loan.csv", low_memory=False)

print("Columns (%d total):" % len(df.columns))
print(df.columns.tolist())
print("\nFirst 20 rows:")
df.head(20)
```

```python
df_norm = df.replace(r'^\s*$', pd.NA, regex=True)
empty_columns = df_norm.columns[df_norm.isna().all()].tolist()
print("Empty columns (%d total):" % len(empty_columns))
print(empty_columns)
empty_columns
```

```python
empty_rows = df[df.isna().all(axis=1)]
print("Empty rows (%d total):" % len(empty_rows))
empty_rows
```

```python
missing_pct = (df_norm.isna().mean() * 100).astype(int)
missing_pct = missing_pct[missing_pct > 0]

for pct in range(100, 0, -1):
    cols = missing_pct[missing_pct == pct].index.tolist()
    print(f"{pct}% missing ({len(cols)} columns):")
    print(cols)
    print()
```

```python
missing_hardship_type = df_norm["hardship_type"].isna().sum()
missing_hardship_type_pct = df_norm["hardship_type"].isna().mean() * 100

print("hardship_type missing values:", missing_hardship_type)
print(f"hardship_type missing percentage: {missing_hardship_type_pct:.2f}%")
```
